# Single Variable Analysis

### Age

```python
age_ascending = pd.DataFrame(df.sort_values(by = 'Age')['Age'].dropna())
print(age_ascending)
```

```python
adult = age_ascending[age_ascending['Age'] > 18]
print(adult)
print(f"adult: {len(adult)/len(age_ascending) * 100}%")

print(f"non-adult: {(1-len(adult)/len(age_ascending)) * 100}%")
```

```python
infants = age_ascending[age_ascending['Age'] <= 3]
print(infants)
print(f"{len(infants)/len(age_ascending) * 100}%")
```

```python
elderly = age_ascending[age_ascending['Age'] >= 60]
print(elderly)
print(f"{len(elderly)/len(age_ascending) * 100}%")
```

#### Analysis: 'pre-adult' accounts for close to 20%. 'Infants' accounts for 4.20%. 'elderly' accounts for 3.64%. The youngest is only 0.42 years old and the oldest is 80 years old.

# Inference

### Sex & Survival

```python
male_survived = pd.DataFrame(male['Survived'])
print(male_survived)

male_survived_count = male_survived.value_counts()
print(male_survived_count)

male_survived_rate = str(109 / len(male_survived) * 100) + '%'
print(male_survived_rate)
```

```python
female_survived = pd.DataFrame(female['Survived'])
print(female_survived)

female_survived_count = female_survived.value_counts()
print(female_survived_count)

female_survived_rate = str(233 / len(female_survived) * 100) + '%'
print(female_survived_rate)
```

#### Analysis: the survival rate of females is much higher than that of males.

### Age & Survival

```python
non_adult = pd.DataFrame(df[df['Age'] < 18])

adult = pd.DataFrame(df[df['Age'] >= 18])

non_adult_survived = non_adult['Survived'].value_counts()

adult_survived = adult['Survived'].value_counts()

print(non_adult_survived)
print(adult_survived)

non_adult_survived_rate = 61/len(non_adult) * 100
adult_survived_rate = 229/len(adult) * 100

print(f'{non_adult_survived_rate}%')
print(f'{adult_survived_rate}%')
```

#### Analysis: The rate of survival of minors is significantly higher than that of adults.

# Conclusion: Female underage is the easiest to survive.
